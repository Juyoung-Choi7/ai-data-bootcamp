```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://news.naver.com/section/103'
res = requests.get(url, timeout=5)

print('상태 코드:', res.status_code)    # 200 = 정상
print('인코딩:   ', res.encoding)
print('HTML 길이:', len(res.text), '자')
print()
print('HTML 앞부분 미리보기:')
print(res.text[:200])
```

```python
html = res.text
soup = BeautifulSoup(html, 'html.parser')
```

#_SECTION_HEADLINE_LIST_f0yey


```python

selector = 'ul[id^="_SECTION_HEADLINE_LIST"]'

headlines = soup.select_one(selector)
for i, li in enumerate(headlines.select("li"), 1):
    title = li.select_one("strong.sa_text_strong").get_text(strip=True)
    np = li.select_one("div.sa_text_press").get_text(strip=True)
    print(f"{i}. {title}. {np}")
```
