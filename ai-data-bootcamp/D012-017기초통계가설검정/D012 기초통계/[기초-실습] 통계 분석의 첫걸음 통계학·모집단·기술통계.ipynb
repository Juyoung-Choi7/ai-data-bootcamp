# 문제 1. 기초통계량 계산

**📘 문제**

- 온라인 쇼핑몰 고객 1,000명을 대상으로 1점부터 10점 사이의 만족도 조사를 실시했습니다.
  이 데이터를 바탕으로 고객 만족도의 **중심 경향성**과 **분포 특성**을 파악해 봅시다.

```python
# 데이터 생성
np.random.seed(2025)
satisfaction = np.random.normal(7.5, 1.5, 1000)
satisfaction = np.clip(satisfaction, 1, 10)
df1 = pd.DataFrame({'score': satisfaction})
```

**📌 아래를 수행해 보세요:**

- 평균, 중앙값, 최빈값, 표준편차를 계산해 봅시다.
- 히스토그램을 그리고 평균과 중앙값 선으로 표시해 봅시다.
- 데이터의 분포 형태를 해석한 후, 마케팅 전략 수립에 어떻게 활용할 수 있을지 생각해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 1] 여기에 코드를 작성해주세요.

mean_val = df1['score'].mean()
median_val = df1['score'].median()
mode_val = df1['score'].mode()[0]   # mode()는 여러 개 나올 수 있어서 [0]으로 첫 값만
std_val = df1['score'].std()

print(f"평균(mean): {mean_val:.2f}")
print(f"중앙값(median): {median_val:.2f}")
print(f"최빈값(mode): {mode_val:.2f}")
print(f"표준편차(std): {std_val:.2f}")

plt.figure(figsize=(8, 5))
plt.hist(df1['score'], bins=20, color='skyblue', edgecolor='black')
plt.axvline(mean_val, color='red', linestyle='--', linewidth=2, label=f'평균: {mean_val:.2f}')
plt.axvline(median_val, color='green', linestyle='-', linewidth=2, label=f'중앙값: {median_val:.2f}')

plt.title('고객 만족도 분포')
plt.xlabel('만족도 점수')
plt.ylabel('빈도')
plt.legend()
plt.show()
```

   - 반대로 최빈값이 평균보다 높은 경우에는 어떤 해석이 가능할까요? 이럴 때 분포가 어떻게 달라질 수 있을까요?

```python
# 데이터를 어떻게 읽을까요?
# [문제 1] 여기에 의견을 작성해주세요.

# 평균과 중앙값의 큰 차이가 없다. 최빈값은 10이지만 
# 평균과 중앙값 모두 낮은 별점들로 인해 점수가 낮아짐
# 그래프는 평균값이 오른쪽으로 살짝 치우친 종모양에 10점 고객의 천장효과 함께 나타남
# 낮은 별점을 준 고객들의 불만사항을 다시 확인하고 보완

# 평균(7.43)과 중앙값(7.42)이 거의 같아 데이터는 대체로 대칭적인 분포다.
# 다만 최빈값이 10으로 나온 건, 설문 척도가 10점 만점이라
# 원래 10점을 넘었을 값들이 전부 10점으로 잘려서(clip) 쌓였기 때문이다(천장 효과).
# 즉 실제로는 "매우 만족한 고객"이 상당수 있는데,
# 현재 척도로는 그들의 만족도 차이를 세밀하게 구분하지 못하고 있다.
# "10점에 몰린 고객층(최고만족 고객)이 많다는 건, 이들을 대상으로 리뷰 작성 요청·추천 프로그램 등 '옹호자(advocate)' 활용 전략을 펼치기 좋다는 뜻이기도 하다. 또한 척도를 세분화(예: 10점 만점 → 15점 만점 등)해서 상위 고객들의 만족도 차이를 더 정밀하게 측정하는 것도 고려할 수 있다."
```

# 문제 2. 평균 vs 중앙값 비교

**📘 문제**

- A지역과 B지역의 연봉 데이터를 비교해 봅시다.
- B 지역에는 일부 고소득자가 포함되어 있습니다.

```python
# 데이터 생성
np.random.seed(2025)

# A 지역: 평균 300, 표준편차 50
region_a = np.random.normal(300, 50, 100)

# B 지역: 일반 소득자 + 극단적인 고소득자 포함
region_b = np.concatenate([
    np.random.normal(280, 40, 95),    # 일반 소득자 95명
    np.random.normal(1500, 200, 5)    # 고소득자 5명
])

df2 = pd.DataFrame({
    'region': ['A'] * 100 + ['B'] * 100,
    'salary': np.concatenate([region_a, region_b])
})
```

**📌 아래를 수행해 보세요:**

- 각 지역의 평균 연봉과 중앙값을 구해 봅시다.
- 지역별 박스플롯(Boxplot)을 그려서 분포의 차이를 시각적으로 확인해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 2] 여기에 코드를 작성해주세요.

# 1. 지역별 평균, 중앙값 계산
summary = df2.groupby('region')['salary'].agg(['mean', 'median'])
print(summary)

# 2. 박스플롯으로 분포 비교
plt.figure(figsize=(8, 5))
sns.boxplot(data=df2, x='region', y='salary')
plt.title('지역별 연봉 분포 비교')
plt.xlabel('지역')
plt.ylabel('연봉')
plt.show()
```

# 문제 3. 이상값과 분산 비교

**📘 문제**

- 1반과 2반의 시험 점수 분포를 비교해 봅시다.
- 2반에는 극단적인 이상값(예: 부정행위로 0점)이 포함되어 있습니다.

```python
# 데이터 생성
np.random.seed(2025)  # 재현 가능성을 위해 시드 고정

# 1반: 평균 75, 표준편차 8
class1 = np.random.normal(75, 8, 30)

# 2반: 일반 학생 + 이상값(0점, 130점)
class2 = np.concatenate([
    np.random.normal(75, 15, 28),  # 일반 학생
    [0, 130]                       # 극단적인 이상값
])

df3 = pd.DataFrame({
    'class': ['1반'] * 30 + ['2반'] * 30,
    'score': np.concatenate([class1, class2])
})
```

**📌 아래를 수행해 보세요:**

1. 각 반의 평균, 표준편차, 최소/최대값을 계산해 봅시다.
2. 박스플롯을 사용하여 두 반의 분포 차이를 시각적으로 비교해 봅시다.
3. 이상값이 통계 지표에 어떤 영향을 주는지 생각해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 3] 여기에 코드를 작성해주세요.

summary = df3.groupby('class')['score'].agg(['mean', 'std', 'min', 'max'])
print(summary)

# 2. 박스플롯으로 분포 비교
plt.figure(figsize=(8, 5))
sns.boxplot(data=df3, x='class', y='score')
plt.title('학급별 점수 분포 비교')
plt.xlabel('학급')
plt.ylabel('점수')
plt.show()
```

# 문제 4. IQR 이상값 제거

**📘 문제**

- 앞의 문제에서 2반에는 **극단적인 이상값(예: 매우 낮은 점수)**이 포함되어 있어
  통계 지표, 특히 평균에 큰 영향을 주는 것을 확인했습니다.
- 이번에는 **IQR(사분위 범위)**을 기준으로 이상값을 제거한 뒤, 이상값 제거 전후의 평균을 비교해 봅시다

<br>

**📌 아래를 수행해 보세요:**

- 2반의 점수 데이터를 대상으로 IQR 기준을 사용해 이상값을 판별하고 제거해 봅시다.
- 이상값 제거 전과 제거 후의 평균을 각각 계산해 봅시다.
- 이상값 제거가 평균에 어떤 영향을 주는지 설명해 봅시다.
- 필요하다면 중앙값과도 비교해 봅시다.

```python
# [문제 4] 여기에 코드를 작성해주세요.

# 1. 2반 데이터만 추출
class2 = df3[df3['class'] == '2반']['score']

# 2. IQR 계산
Q1 = class2.quantile(0.25)
Q3 = class2.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"이상값 판별 범위: {lower_bound:.2f} ~ {upper_bound:.2f}")

# 3. 이상값 제거
class2_clean = class2[(class2 >= lower_bound) & (class2 <= upper_bound)]

# 4. 제거 전/후 비교
print(f"\n[제거 전] 평균: {class2.mean():.2f}, 중앙값: {class2.median():.2f}, 개수: {len(class2)}")
print(f"[제거 후] 평균: {class2_clean.mean():.2f}, 중앙값: {class2_clean.median():.2f}, 개수: {len(class2_clean)}")
print(f"\n제거된 이상값: {class2[~class2.index.isin(class2_clean.index)].tolist()}")
```

   - 두 지표를 함께 보면 어떤 장점이 있을까요?

```python
# 데이터를 어떻게 읽을까요?
# [문제 4] 여기에 의견을 작성해주세요.

# 이상값(0점, 130점) 제거 전후 평균은 73.73 → 74.35로 0.62점만 차이 났다.
# 평균이 중앙값(73.00)에 더 가까워지긴 했지만, 그 폭은 미미했다.
# 이는 두 이상값이 각각 반대 방향(매우 낮음/매우 높음)으로 작용해
# 서로의 영향을 상당 부분 상쇄시켰기 때문이다.
# 즉 이상값이 "존재"했음에도 평균을 크게 왜곡시키지는 않은 특수한 사례다.

# IQR은 데이터를 4등분해서 중간 50%(Q1~Q3)의 퍼짐 정도만 보기 때문에,
# 극단값 자체에 영향을 받지 않고 이상값을 정의할 수 있다.
# 반면 평균이나 표준편차로 기준을 잡으면, 그 기준 자체가 이미
# 이상값의 영향을 받은 값이므로 "이상값을 판단하는 도구가
# 이상값에 의해 왜곡되는" 순환 문제가 생긴다.
# 그래서 IQR처럼 중앙값 계열의 강건한(robust) 지표를 기준으로 삼는 것이 안전하다.

# 이번 사례처럼 0점(백지 제출 등 측정 오류 가능성)이나 130점(만점 초과,
# 데이터 입력 오류 가능성)처럼 "실제로 잘못된 값"일 가능성이 있다면 제거가 타당하다.
# 반면 만약 130점이 오류가 아니라 실제로 특별 가산점을 받은 학생의
# 정당한 점수라면, 이를 제거하는 것은 오히려 중요한 정보를 왜곡하는 것이다.
# 따라서 이상값 제거는 통계적 기준(IQR)뿐 아니라,
# 그 값이 발생한 맥락(데이터 오류인지, 실제 현상인지)을 함께 판단해야 한다.

# 평균은 모든 값을 반영하기 때문에 이상값이 없는 데이터에서는
# 전체 경향을 잘 대표한다. 하지만 이번처럼 극단값이 섞여 있으면
# 평균은 왜곡될 여지가 있다(다만 이번 사례는 상쇄로 인해 왜곡이 작았다).
# 중앙값은 극단값에 영향받지 않아 "전형적인 값"을 보여주는 데 강하다.
# 실제로 이번 데이터에서는 이상값 제거 여부와 관계없이 중앙값이
# 73.00으로 전혀 변하지 않아, 중앙값의 강건성을 잘 보여준다.
# 따라서 두 지표를 함께 보면, 평균과 중앙값의 차이 자체가
# "데이터에 이상값이 얼마나 영향을 주고 있는지"를 알려주는 신호가 된다.
```

# 문제 9. 정규분포와 표준편차

**📘 문제**

- 정규분포를 따르는 점수 데이터에서 다음을 확인해 봅시다:
  - 평균 ±1σ, 평균 ±2σ 범위에 속한 값들의 비율은 실제로 얼마나 되는가?
  - 이 결과가 이론적인 68-95-99 법칙과 얼마나 유사한지 비교해 봅시다.

<br>

**📌 아래를 수행해 보세요:**

2. 평균을 중심으로 ±1σ, ±2σ 범위에 포함되는 데이터의 비율을 계산해 봅시다.
3. 결과를 바탕으로 68-95-99 법칙과 비교해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 9] 여기에 코드를 작성해주세요.

# [문제 9] 여기에 코드를 작성해주세요.

import numpy as np

# 1. 정규분포 데이터 생성 (평균 70, 표준편차 10, 1000개)
np.random.seed(42)
scores = np.random.normal(70, 10, 1000)

mean = scores.mean()
std = scores.std()

print(f"평균: {mean:.2f}, 표준편차: {std:.2f}")

# 2. ±1σ, ±2σ 범위에 속한 데이터 비율 계산
within_1std = ((scores >= mean - 1*std) & (scores <= mean + 1*std)).mean() * 100
within_2std = ((scores >= mean - 2*std) & (scores <= mean + 2*std)).mean() * 100

print(f"\n±1σ 범위 ({mean-std:.2f} ~ {mean+std:.2f}) 안의 비율: {within_1std:.2f}%")
print(f"±2σ 범위 ({mean-2*std:.2f} ~ {mean+2*std:.2f}) 안의 비율: {within_2std:.2f}%")

# 3. 이론값(68-95-99.7 법칙)과 비교
print(f"\n[이론값] ±1σ: 68%, ±2σ: 95%")
print(f"[실제값] ±1σ: {within_1std:.2f}%, ±2σ: {within_2std:.2f}%")
print(f"[차이]   ±1σ: {within_1std-68:.2f}%p, ±2σ: {within_2std-95:.2f}%p")
```

# 문제 10. 표본 평균 실험

**📘 문제**

- 큰 모집단에서 표본을 여러 번 추출하여 표본 평균의 분포를 비교해봅시다.

<br>

**📌 아래를 수행해 보세요:**

1. 평균이 100, 표준편차가 15인 모집단을 생성합니다.
2. 다음 두 조건에서 각각 1000번씩 표본 평균을 구합니다:
   - 표본 크기 5
   - 표본 크기 50
3. 두 조건의 **표본 평균 분포(histogram)**를 비교합니다.

```python
# 아래를 수행해 보세요:
# [문제 10] 여기에 코드를 작성해주세요.

# [문제 10] 여기에 코드를 작성해주세요.

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# 1. 모집단 생성 (평균 100, 표준편차 15)
population = np.random.normal(100, 15, 100000)  # 충분히 큰 모집단

# 2. 표본 크기 5로 1000번 표본 추출 → 표본 평균 1000개
sample_means_5 = []
for _ in range(1000):
    sample = np.random.choice(population, size=5, replace=False)
    sample_means_5.append(sample.mean())

# 3. 표본 크기 50으로 1000번 표본 추출 → 표본 평균 1000개
sample_means_50 = []
for _ in range(1000):
    sample = np.random.choice(population, size=50, replace=False)
    sample_means_50.append(sample.mean())

sample_means_5 = np.array(sample_means_5)
sample_means_50 = np.array(sample_means_50)

# 4. 두 표본평균 분포의 평균/표준편차 비교
print(f"[표본 크기 5]  평균: {sample_means_5.mean():.2f}, 표준편차: {sample_means_5.std():.2f}")
print(f"[표본 크기 50] 평균: {sample_means_50.mean():.2f}, 표준편차: {sample_means_50.std():.2f}")

# 5. 히스토그램으로 비교 시각화
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].hist(sample_means_5, bins=30, color='skyblue', edgecolor='black')
axes[0].set_title('표본 크기 5의 표본평균 분포')
axes[0].set_xlabel('표본평균')
axes[0].set_xlim(60, 140)

axes[1].hist(sample_means_50, bins=30, color='salmon', edgecolor='black')
axes[1].set_title('표본 크기 50의 표본평균 분포')
axes[1].set_xlabel('표본평균')
axes[1].set_xlim(60, 140)

plt.tight_layout()
plt.show()
```

# 문제 11. 단순 선형 회귀 분석

**📘 문제**

- 공부 시간과 성적 간의 관계를 회귀 분석으로 확인해봅시다.

```python
# 데이터 생성
from sklearn.linear_model import LinearRegression

np.random.seed(505)
study_time = np.random.normal(5, 2, 100)
score = 50 + study_time * 8 + np.random.normal(0, 5, 100)
df11 = pd.DataFrame({'study_time': study_time, 'score': score})
```

**📌 아래를 수행해 보세요:**

1. 공부 시간(`study_time`)과 성적(`score`)간의 관계를 단순 선형 회귀모델로 분석합니다.
2. 회귀 모델을 학습하고, 기울기(계수)와 절편을 출력해 봅시다.
3. 산점도 위에 회귀선을 시각화해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 11] 여기에 코드를 작성해주세요.

# [문제 11] 여기에 코드를 작성해주세요.

from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# 1. 회귀모델 학습
X = df11[['study_time']]
y = df11['score']

model = LinearRegression()
model.fit(X, y)

# 2. 기울기(계수)와 절편 출력
slope = model.coef_[0]
intercept = model.intercept_

print(f"기울기(계수): {slope:.2f}")
print(f"절편: {intercept:.2f}")
print(f"회귀식: score = {intercept:.2f} + {slope:.2f} * study_time")

# 3. R² (결정계수) 확인
r_squared = model.score(X, y)
print(f"R² (결정계수): {r_squared:.4f}")

# 4. 산점도 + 회귀선 시각화
plt.figure(figsize=(8, 6))
plt.scatter(df11['study_time'], df11['score'], alpha=0.6, label='실제 데이터')

y_pred = model.predict(X)
plt.plot(df11['study_time'], y_pred, color='red', linewidth=2, label='회귀선')

plt.title('공부 시간과 성적의 관계 (단순 선형 회귀)')
plt.xlabel('공부 시간 (study_time)')
plt.ylabel('성적 (score)')
plt.legend()
plt.show()
```

# 문제 12. A/B 테스트 해석

**📘 문제**

- 두 그룹의 전환율 차이를 계산하고 해석해봅시다.

```python
# 데이터 생성
np.random.seed(606)
group_a = np.random.binomial(1, 0.12, 1000)
group_b = np.random.binomial(1, 0.15, 1000)
```

**📌 아래를 수행해 보세요:**

1. A 그룹과 B 그룹의 전환 데이터를 살펴봅니다.
2. 각 그룹의 전환율을 계산하고, 차이를 확인합니다.
3. 이 차이가 우연인지, 의미 있는 차이인지 생각해 봅시다.

```python
# 아래를 수행해 보세요:
# [문제 12] 여기에 코드를 작성해주세요.

from scipy import stats

# 1. 각 그룹의 전환율 계산
rate_a = group_a.mean()
rate_b = group_b.mean()

print(f"A 그룹 전환율: {rate_a:.4f} ({rate_a*100:.2f}%)")
print(f"B 그룹 전환율: {rate_b:.4f} ({rate_b*100:.2f}%)")
print(f"전환율 차이: {(rate_b - rate_a)*100:.2f}%p")

# 2. 이 차이가 통계적으로 유의미한지 검정 (독립표본 t검정)
t_stat, p_value = stats.ttest_ind(group_a, group_b)

print(f"\nt-통계량: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

if p_value < 0.05:
    print("→ p-value < 0.05, 두 그룹의 전환율 차이는 통계적으로 유의미하다.")
else:
    print("→ p-value >= 0.05, 두 그룹의 전환율 차이가 우연일 가능성을 배제할 수 없다.")
```

**🧠 데이터를 어떻게 읽을까요?**

1. 두 그룹의 전환율 차이는 얼마나 되나요?

3. 이 실험 결과로 어떤 결정을 내릴 수 있을까요?
4. A/B 테스트는 현실에서 어떻게 쓰일까요?

```python
# 데이터를 어떻게 읽을까요?
# [문제 12] 여기에 의견을 작성해주세요.

# 1. 두 그룹의 전환율 차이는 얼마나 되나요?
# A 그룹 13.10%, B 그룹 13.30%로, 차이는 0.20%p에 불과하다.
# 원래 데이터 생성 시 설계된 실제 전환율은 A=12%, B=15%로
# 3%p 차이가 나도록 만들어졌지만, 실제 표본(1000명씩)에서는
# 우연히 이보다 훨씬 작은 차이만 관측되었다.

# 2. 차이가 '의미 있다'는 건 무엇을 뜻하나요?
# '의미 있다(통계적으로 유의미하다)'는 것은, 관측된 차이가
# 단순히 우연(표본 추출 과정의 무작위성)으로 발생했을 가능성이
# 낮다는 뜻이다. p-value가 그 기준을 제공하는데,
# 이번 검정에서는 p-value가 0.8950으로 매우 커서,
# "이 정도 차이는 우연히도 아주 흔하게 나타날 수 있는 수준"이라는 뜻이다.
# 즉 지금 관측된 0.20%p 차이는 통계적으로 유의미하지 않다.

# 3. 이 실험 결과로 어떤 결정을 내릴 수 있을까요?
# 현재 결과만으로는 B안이 A안보다 낫다고 결론 내릴 수 없다.
# 흥미로운 점은, 실제로는 B의 진짜 전환율(15%)이 A(12%)보다
# 높게 설계되어 있었음에도, 이번 표본에서는 우연히 그 차이가
# 거의 드러나지 않았다는 것이다. 이는 "실제 효과가 존재해도,
# 특정 한 번의 실험/표본에서는 그 효과가 우연에 가려져
# 보이지 않을 수 있다"는 중요한 교훈을 준다.
# 따라서 이 결과만으로 "차이가 없다"고 단정하기보다는,
# 표본 크기를 늘리거나 실험을 반복해 재검증하는 것이 안전하다.

# 4. A/B 테스트는 현실에서 어떻게 쓰일까요?
# - 웹사이트/앱의 버튼 색상, 문구, 레이아웃 변경 시
#   실제로 전환율(구매, 가입, 클릭 등)이 개선되는지 검증
# - 마케팅 이메일 제목이나 광고 문구의 효과 비교
# - 가격 정책 변경(할인율 등)이 매출에 미치는 영향 검증
# - 신약이나 치료법의 효과를 대조군과 비교하는 임상시험
# 공통적으로, 눈에 보이는 차이만으로 성급히 결정하지 않고
# 통계적 유의성 검정을 거쳐야 신뢰할 수 있는 의사결정이 가능하다.
```
