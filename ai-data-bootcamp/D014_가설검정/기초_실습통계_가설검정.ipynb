# 문제 1. 게임 아이템 뽑기 확률, 정말 10%일까? (난이도: 하)

**📘 문제**

- 어떤 게임 개발사에서 새로운 아이템의 뽑기 확률이 **10%**로 설정되었다고 주장하고 있습니다.

- 그런데 유저 커뮤니티에서는 "실제 확률은 10%보다 낮은 것 같다"는 의혹이 제기되었습니다.

- 이를 확인하기 위해 한 유저가 아이템을 **200번 뽑았고, 그중 12번 성공**했습니다.

- 이번 실습에서는 **이항검정(Binomial Test)**으로 이 결과가 개발사의 주장을 기각할 만큼 유의미한지 확인하고,
  시뮬레이션을 통해 **p-값이 무엇을 의미하는지**를 직접 눈으로 확인해 봅니다.

```python
# 문제 설정
n = 200  # 총 시행 횟수
x = 12   # 관찰된 성공 횟수
p = 0.1  # 귀무가설에서의 성공 확률

print(f"관찰 결과: {n}번 시도 중 {x}번 성공 (관찰 성공률 {x/n:.1%})")
```

**📌 아래를 수행해 보세요:**

- 귀무가설(H₀)과 대립가설(H₁)을 세워 봅시다.

- 이항검정을 수행하여 p-값을 구하고, 유의수준 0.05를 기준으로 결론을 내려봅시다.

- **"만약 개발사의 주장대로 실제 확률이 정말 10%라면"** 200번 뽑기에서 성공 횟수가 어떻게 분포하는지 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 우리가 관찰한 '12번 성공'이 얼마나 희귀한 일인지 관찰해 봅시다.

```python
# [문제 1] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 새로운 아이템 뽑기 확률은 10%이다.
# H₁ (대립가설): 새로운 아이템 뽑기 확률은 10%보다 낮다.
```

```python
# [문제 1] Q2. 이항검정을 수행하고 p-값을 계산해봅시다.
# '확률이 더 낮은 것 같다'는 주장을 검증하려면 alternative 인수를 어떻게 설정해야 할까요?

# 여기에 코드를 작성해주세요.
from scipy import stats

result = stats.binomtest(x, n, p, alternative='less')
p_value = result.pvalue
print(f"p-값: {p_value:.6f}")

from math import comb

# 정확히 12번 성공할 확률 (이항분포 PMF)
prob_exact_12 = comb(200, 12) * (p**12) * ((1-p)**(200-12))
print(f"정확히 12번 성공할 확률: {prob_exact_12:.6f}")

# p-값 = 0번부터 12번까지 성공할 확률을 모두 더한 값 (왼쪽 꼬리)
p_value_manual = 0
for k in range(0, 13):  # 0부터 12까지
    prob_k = comb(200, 12 if False else k) * (p**k) * ((1-p)**(200-k))
    p_value_manual += prob_k

print(f"직접 계산한 p-값: {p_value_manual:.6f}")
```

```python

# [문제 1] Q3. 유의수준 0.05를 기준으로 통계적 결론을 내리는 코드를 작성해봅시다.
alpha = 0.05

# 여기에 코드를 작성해주세요.

if p_value < alpha:
    print(f"p-값({p_value:.6f}) < 유의수준({alpha}) → 귀무가설(H₀) 기각")
    print("→ '뽑기 확률이 10%보다 낮다'는 유저 커뮤니티의 의혹이 통계적으로 유의미합니다.")
else:
    print(f"p-값({p_value:.6f}) >= 유의수준({alpha}) → 귀무가설(H₀) 기각 실패")
    print("→ 현재 데이터로는 '확률이 10%보다 낮다'고 통계적으로 결론짓기 어렵습니다.")
```

```python
# [문제 1] Q4. H₀가 사실(실제 확률 = 10%)이라고 가정하고,
# 200번 뽑기를 10,000번 반복하는 시뮬레이션을 수행해봅시다.
# 힌트: numpy의 이항분포 랜덤 함수(np.random.binomial)를 사용해보세요.

# 여기에 코드를 작성해주세요.

np.random.seed(2025)
n_simulations = 10000

sim_results = np.random.binomial(n=200, p=0.1, size=n_simulations)

print(f"시뮬레이션 횟수: {n_simulations}")
print(f"시뮬레이션 성공 횟수의 평균: {sim_results.mean():.2f} (이론값: {200*0.1})")
print(f"12번 이하로 성공한 시뮬레이션 비율: {(sim_results <= 12).mean():.4f}")
```

```python
# [문제 1] Q5. 시뮬레이션 결과를 히스토그램으로 그려봅시다.
# 실제 관찰값(12번 성공)을 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.

plt.figure(figsize=(9, 5))
sns.histplot(sim_results, bins=range(sim_results.min(), sim_results.max()+2), discrete=True)

plt.axvline(12, color='red', linestyle='--', linewidth=2, label='실제 관찰값: 12번 성공')
plt.axvline(sim_results.mean(), color='green', linestyle='-', linewidth=2, 
            label=f'시뮬레이션 평균: {sim_results.mean():.1f}')

plt.title("H₀(성공확률 10%) 가정 하의 200번 뽑기 성공 횟수 분포 (10,000회 시뮬레이션)")
plt.xlabel("성공 횟수")
plt.ylabel("빈도")
plt.legend()
plt.show()
```

**🧠 데이터를 어떻게 읽을까요?**

- Q2에서 구한 p-값은 시뮬레이션 그래프의 어느 부분에 해당하며, 무엇을 의미하나요?

- 시뮬레이션 그래프를 볼 때, 우리가 관찰한 '12회 성공'은 개발사의 주장이 맞다는 가정 하에 흔한 일인가요, 드문 일인가요?

- 이 시뮬레이션 경험을 통해 "p-값이 작으면 귀무가설을 기각한다"는 규칙을 친구에게 어떻게 더 쉽게 설명할 수 있을까요?

- 만약 200번이 아니라 1,000번을 뽑아 같은 성공률(6%)이 나왔다면, p-값은 어떻게 달라질까요?

```python
# [문제 1] 데이터를 어떻게 읽을까요?
# 여기에 의견을 작성해주세요.

# p-값은 빨간선으로부터 왼쪽 구간의 넓이를 의미합니다.
# 이는 귀무가설 (아이템뽑기 확률이 10%) 하에서 
# 아이템이 200번 중 12번 이하로 뽑힐 확률을 의미합니다.

# p-값은 0.0320으로 이는 유의수준으로 설정한 0.05보다 낮습니다.
# 이는 개발사의 주장이 맞다는 가정하에 드문일이라는 것을 의미합니다.

# p-값이 작다는 것은 귀무가설하에서 관찰된 사건이 
# 우연히 일어날 확률이 매우 작다는 뜻입니다.
# 따라서 귀무가설이 기각됩니다.

# p-값은 더 작아집니다.
# 이유: 표본 크기가 커질수록 표본분포의 표준오차(SE)가 작아져
# (SE = sqrt(n*p*(1-p))가 커지긴 하지만 상대적 변동성은 줄어듦),
# 같은 비율의 편차라도 귀무가설 분포 안에서 더 극단적인 위치로 나타나기 때문입니다.
# 즉, 관찰 횟수가 많을수록 "우연히 그렇게 나왔을 가능성"이 줄어들어
# 같은 비율이라도 통계적 증거력이 더 강해집니다.
```

# 문제 2. 과자 한 봉지의 중량은 150g이 맞을까? (난이도: 하)

**📘 문제**

- 한 식품 공장에서 생산하는 과자 한 봉지의 **목표 중량은 150g**입니다.

- 품질관리팀은 생산 공정이 목표 중량을 잘 맞추고 있는지 확인하기 위해,
  생산된 과자 **30봉지를 무작위로 추출**하여 무게를 측정했습니다.

- 이번 실습에서는 **일표본 t-검정(One-sample t-Test)**으로 표본 평균이 목표 중량과 유의미하게 다른지 검증하고,
  **"평균이 정확히 150g이라면"** 어떤 표본 평균들이 나올 수 있는지 시뮬레이션으로 확인해 봅니다.

```python
# 데이터 생성 (과자 30봉지의 무게 측정값)
np.random.seed(42)
sample_weights = np.random.normal(loc=151.5, scale=2, size=30)
pop_mean = 150  # 목표 중량 (모평균)

# 표본 분포 시각화
sns.histplot(sample_weights, bins=10, kde=True)
plt.axvline(x=pop_mean, color='gray', linestyle='--', label='목표 중량 150g')
plt.title("추출한 과자 30봉지의 무게 분포")
plt.xlabel("무게 (g)")
plt.legend()
plt.show()
```

**📌 아래를 수행해 보세요:**

- 귀무가설(H₀)과 대립가설(H₁)을 세워 봅시다.

- 일표본 t-검정을 수행하여 t-통계량과 p-값을 구하고, 유의수준 0.05를 기준으로 결론을 내려봅시다.

- **"평균 무게가 정확히 150g이라면"** 30개짜리 표본의 평균이 어떻게 분포하는지 10,000번 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 실제 관찰된 표본 평균이 그 분포에서 어디에 위치하는지 관찰해 봅시다.

```python
# [문제 2] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 과자 한 봉지의 평균 중량은 150g이다.
# H₁ (대립가설): 과자 한 봉지의 평균 중량은 150g이 아니다.
```

```python
# [문제 2] Q2. 일표본 t-검정을 수행하여 t-통계량과 p-값을 구해봅시다.

# 여기에 코드를 작성해주세요.
from scipy import stats

t_stat, p_value = stats.ttest_1samp(sample_weights, popmean=150)

print(f"t-통계량: {t_stat:.4f}")
print(f"p-값: {p_value:.4f}")
```

```python
# [문제 2] Q3. 유의수준 0.05를 기준으로 통계적 결론을 내리는 코드를 작성해봅시다.
# alpha = 0.05

# 여기에 코드를 작성해주세요.

alpha = 0.05

if p_value < alpha:
    print(f"p-값({p_value:.4f}) < 유의수준({alpha}) → 귀무가설(H₀) 기각")
    print("→ 과자 봉지의 평균 중량은 150g과 통계적으로 유의미하게 다릅니다.")
else:
    print(f"p-값({p_value:.4f}) >= 유의수준({alpha}) → 귀무가설(H₀) 기각 실패")
    print("→ 현재 데이터로는 평균 중량이 150g과 다르다고 말하기 어렵습니다.")
```

```python
# [문제 2] Q4. H₀가 사실(평균 = 150g)이라고 가정하고, 30개짜리 표본을 10,000번 뽑아
# 각 표본의 평균을 리스트에 저장해봅시다.
# 힌트: 표준편차는 우리가 가진 표본의 표준편차를 사용한다고 가정하고, for 반복문을 사용해보세요.

# 여기에 코드를 작성해주세요.

sample_std = sample_weights.std(ddof=1)  # 표본표준편차 (자유도 보정)

np.random.seed(2025)
sim_means = []
for i in range(10000):
    sim_sample = np.random.normal(loc=150, scale=sample_std, size=30)
    sim_means.append(sim_sample.mean())

sim_means = np.array(sim_means)

print(f"시뮬레이션 표본평균들의 평균: {sim_means.mean():.4f} (이론값: 150)")
print(f"시뮬레이션 표본평균들의 표준편차: {sim_means.std():.4f}")
```

```python
# [문제 2] Q5. 시뮬레이션으로 얻은 표본 평균들의 분포를 히스토그램으로 그려봅시다.
# 실제 관찰된 표본 평균을 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.
observed_mean = sample_weights.mean()

plt.figure(figsize=(9, 5))
sns.histplot(sim_means, bins=40, kde=True)

plt.axvline(observed_mean, color='red', linestyle='--', linewidth=2,
            label=f'실제 관찰된 표본평균: {observed_mean:.2f}')
plt.axvline(150, color='green', linestyle='-', linewidth=2, label='귀무가설 평균: 150')

plt.title("H₀(평균=150g) 가정 하의 표본평균 분포 (10,000회 시뮬레이션)")
plt.xlabel("표본 평균 (g)")
plt.legend()
plt.show()
```

- Q5의 히스토그램은 무엇을 나타내나요? 이 분포의 모양이 정규분포와 비슷한 이유는 무엇일까요? (힌트: 4장에서 배운 중심극한정리)

- 우리가 실제로 관찰한 표본 평균(세로선)은 이 분포에서 흔하게 나타나는 값인가요? Q2의 p-값과 이 시각적 위치는 어떻게 관련되나요?

- 만약 p-값이 0.001이었다면, 세로선은 그래프의 어디쯤에 위치할 것으로 예상되나요?

- 이 결과는 공장 입장에서 어떤 조치를 취해야 함을 시사할까요? (목표보다 무겁게 나오는 것은 어떤 손실일까요?)

```python
# [문제 2] 데이터를 어떻게 읽을까요?
# 여기에 의견을 작성해주세요.

# 히스토그램은 귀무가설 하에서 30개짜리 표본을 10,000번 추출해
# 각각의 표본평균을 모은 분포입니다.
# 분포가 정규분포에 가까운 이유는 표본 크기(n=30)가 충분히 크기 때문에
# 중심극한정리가 성립하기 때문입니다.
# (반복 횟수 10,000번은 분포를 더 매끄럽게 보여줄 뿐, 
# 정규분포 모양 자체를 만드는 요인은 아닙니다.)

# 우리가 실제 관찰한 표본평균은 오른쪽에 매우 치우쳐 있습니다.
# 이를 통해 귀무가설 하에서는 우리가 관찰한 결과가
# 매우 드물게 발생함을 알 수 있습니다.
# 이번 검정은 양측검정이므로 
# p-값은 빨간선의 오른쪽 넓이의 두배와 같습니다.

# 만약 p-값이 0.001이었다면, 관찰된 표본평균(세로선)은
# 지금(151.12g / p-값: 0.0019)보다 분포의 중심(150g)에서 더 멀리 떨어진 위치,
# 즉 오른쪽 꼬리의 더 바깥쪽에 위치했을 것입니다.
# p-값이 작다는 것은 그만큼 극단적인(드문) 결과라는 뜻이기 때문입니다.

# 목표(150g)보다 무겁게 나온다는 것은 봉지마다 원료(과자)가
# 계획보다 더 많이 들어간다는 뜻이며, 이는 다음과 같은 손실로 이어집니다.
#
# 1) 원가 손실: 봉지 하나당 평균 1.5g씩 더 들어가면,
#    생산량이 많을수록 누적된 원료비 손실이 커집니다.
# 2) 표시 중량 문제: 포장에 표시된 중량(150g)보다 실제로 더 들어가는 것 자체는
#    소비자 클레임 문제는 아니지만, 회사 입장에서는 "의도치 않게 더 주는" 비효율입니다.
# 3) 공정 신뢰성 문제: 목표치와 실제 값이 유의미하게 다르다는 것은
#    생산 설비나 계량 공정이 정확히 캘리브레이션되어 있지 않을 가능성을 시사합니다.
#
# 따라서 공장은 계량 설비를 재점검하고 공정을 재조정하여
# 평균 중량을 목표치(150g)에 맞추는 개선 조치가 필요합니다.
```

# 문제 3. 어떤 온라인 학습 방식이 더 효과적일까? (난이도: 중)

**📘 문제**

- 한 교육 기업에서 두 가지 다른 온라인 학습 방식(A, B)을 개발했습니다.

- 방식 A가 방식 B보다 학생들의 성적 향상에 더 효과적인지 알아보기 위해,
  두 그룹의 학생들에게 각각 다른 방식으로 한 달간 학습시킨 후 시험을 보게 했습니다.

- 이번 실습에서는 **이표본 t-검정(Two-sample t-Test)**으로 두 방식의 평균 점수 차이를 검증합니다.
  단, t-검정의 기본 가정인 **정규성**과 **등분산성**을 먼저 확인해야 합니다.

- 또한 **"두 방식의 효과가 완전히 똑같다면"** 우연히 나타날 수 있는 점수 차이의 범위를 시뮬레이션으로 확인해 봅니다.

```python
# 데이터 생성 (두 학습 방식의 시험 점수)
np.random.seed(0)
group_a_scores = np.random.normal(loc=85, scale=8, size=50)
group_b_scores = np.random.normal(loc=80, scale=7, size=50)

df_scores = pd.DataFrame({'점수': np.concatenate([group_a_scores, group_b_scores]),
                          '학습방식': ['A']*50 + ['B']*50})

# 두 그룹의 점수 분포 시각화
sns.boxplot(x='학습방식', y='점수', data=df_scores)
plt.title("학습 방식에 따른 시험 점수 분포")
plt.show()
```

**📌 아래를 수행해 보세요:**

- 두 그룹에 대해 **정규성 검정(Shapiro-Wilk test)**을 수행해 봅시다. (H₀: 데이터는 정규분포를 따른다.)

- **등분산성 검정(Levene's test)**을 수행해 봅시다. (H₀: 두 그룹의 분산은 같다.)

- 가정 검토 결과에 따라 `equal_var` 인수를 설정하여 **이표본 t-검정**을 수행해 봅시다.

- 두 그룹 데이터를 모두 합친 뒤 무작위로 다시 나누는 방식으로, **"두 방식의 효과가 같을 때"**의 평균 차이 분포를 시뮬레이션해 봅시다.

```python
# [문제 3] Q1. 두 그룹에 대해 각각 정규성 검정(Shapiro-Wilk test)을 수행하고 p-값을 출력해봅시다.
# H₀: 데이터는 정규분포를 따른다.

# 여기에 코드를 작성해주세요.

from scipy import stats

alpha = 0.05

stat_a, p_a = stats.shapiro(group_a_scores)
stat_b, p_b = stats.shapiro(group_b_scores)

print(f"A그룹 Shapiro-Wilk p-값: {p_a:.4f}")
print(f"B그룹 Shapiro-Wilk p-값: {p_b:.4f}")

result_a = "정규분포를 따른다고 볼 수 있음" if p_a >= alpha else "정규분포를 따른다고 보기 어려움"
result_b = "정규분포를 따른다고 볼 수 있음" if p_b >= alpha else "정규분포를 따른다고 보기 어려움"

print(f"→ A그룹: p-값({p_a:.4f}) {'>=' if p_a >= alpha else '<'} 유의수준({alpha}) → {result_a}")
print(f"→ B그룹: p-값({p_b:.4f}) {'>=' if p_b >= alpha else '<'} 유의수준({alpha}) → {result_b}")
```

```python
# [문제 3] Q2. 두 그룹에 대해 등분산성 검정(Levene's test)을 수행하고 p-값을 출력해봅시다.
# H₀: 두 그룹의 분산은 같다.

# 여기에 코드를 작성해주세요.

stat_levene, p_levene = stats.levene(group_a_scores, group_b_scores)

print(f"Levene's test p-값: {p_levene:.4f}")

levene_result = "두 그룹의 분산이 같다고 볼 수 있음 (등분산 가정 충족)" if p_levene >= alpha else "두 그룹의 분산이 다르다고 볼 수 있음 (등분산 가정 불충족)"

print(f"→ p-값({p_levene:.4f}) {'>=' if p_levene >= alpha else '<'} 유의수준({alpha}) → {levene_result}")
```

```python
# [문제 3] Q3. 이표본 t-검정을 수행하여 t-통계량과 p-값을 구해봅시다.
# H₀: 두 그룹의 평균은 같다. / H₁: 두 그룹의 평균은 다르다.
# 등분산성 가정을 만족했는지 여부에 따라 equal_var 인수를 설정해보세요.

# 여기에 코드를 작성해주세요.

equal_var_assumption = p_levene >= alpha

t_stat, p_value = stats.ttest_ind(group_a_scores, group_b_scores, equal_var=equal_var_assumption)

assumption_text = "충족 → 일반 t-test 적용" if equal_var_assumption else "불충족 → Welch's t-test 적용"

print(f"등분산 가정: {assumption_text}")
print(f"t-통계량: {t_stat:.4f}")
print(f"p-값: {p_value:.4f}")

if p_value < alpha:
    print(f"→ p-값({p_value:.4f}) < 유의수준({alpha}) → 귀무가설(H₀) 기각")
    print("→ A방식과 B방식의 평균 점수는 통계적으로 유의미하게 다릅니다.")
else:
    print(f"→ p-값({p_value:.4f}) >= 유의수준({alpha}) → 귀무가설(H₀) 기각 실패")
    print("→ 현재 데이터로는 두 방식의 평균 점수가 다르다고 말하기 어렵습니다.")
```

```python
# [문제 3] Q4. H₀가 사실(두 그룹의 평균이 같다)이라고 가정하고 시뮬레이션을 10,000번 수행해봅시다.
# 두 그룹의 데이터를 모두 합친 뒤(np.concatenate), 매 반복마다 비복원추출로 50개(가상 A그룹)와
# 나머지 50개(가상 B그룹)로 나누어 두 그룹의 평균 차이를 계산하고 리스트에 저장해보세요.
# 힌트: np.random.permutation 을 사용해보세요.

# 여기에 코드를 작성해주세요.

combined = np.concatenate([group_a_scores, group_b_scores])
n_a = len(group_a_scores)

np.random.seed(2025)
sim_diffs = []
for i in range(10000):
    shuffled = np.random.permutation(combined)
    sim_a = shuffled[:n_a]
    sim_b = shuffled[n_a:]
    sim_diffs.append(sim_a.mean() - sim_b.mean())

sim_diffs = np.array(sim_diffs)

print(f"시뮬레이션 평균 차이의 평균: {sim_diffs.mean():.4f} (이론값: 0)")
print(f"시뮬레이션 평균 차이의 표준편차: {sim_diffs.std():.4f}")
```

```python
# [문제 3] Q5. 시뮬레이션으로 얻은 평균 차이의 분포를 히스토그램으로 그려봅시다.
# 실제 관찰된 평균 차이(A그룹 평균 - B그룹 평균)를 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.

observed_diff = group_a_scores.mean() - group_b_scores.mean()

plt.figure(figsize=(9, 5))
sns.histplot(sim_diffs, bins=40, kde=True)

plt.axvline(observed_diff, color='red', linestyle='--', linewidth=2,
            label=f'실제 관찰된 평균 차이: {observed_diff:.2f}')
plt.axvline(0, color='green', linestyle='-', linewidth=2, label='귀무가설: 차이 없음(0)')

plt.title("H₀(두 그룹 평균 동일) 가정 하의 평균 차이 분포 (10,000회 순열검정)")
plt.xlabel("평균 차이 (A - B)")
plt.legend()
plt.show()
```

**🧠 데이터를 어떻게 읽을까요?**

- 정규성 검정과 등분산성 검정의 p-값은 각각 어떻게 해석해야 하나요? 이 결과는 이표본 t-검정을 사용하는 데 문제가 없음을 보여주나요?

- Q5의 히스토그램 중심이 0에 가까운 이유는 무엇일까요?

- 실제 관찰된 평균 차이(세로선)는 "두 방식의 효과가 같다"고 가정했을 때 우연히 나타날 수 있는 범위 안에 있나요, 벗어나나요?

- 이 시각적 판단과 Q3의 p-값은 어떤 관계가 있나요?

- 통계적으로 유의미한 차이가 있다면, 교육 기업은 이 결과를 어떻게 활용할 수 있을까요?

```python
# [문제 3] 데이터를 어떻게 읽을까요?
# 여기에 의견을 작성해주세요.

# 정규성 검정과 등분산성 검정의 귀무가설은
# 각각 정규성을 만족한다와 등분산성을 만족한다입니다.
# 따라서 p-값이 클 때 귀무가설을 기각하지 않고
# 이를 정규성과 등분산성을 만족한다고 해석합니다.
# 위 결과는 정규성을 만족하고 등분산성을 만족하지 않습니다.
# 따라서 모수검정이자 등분산성을 만족하지 않을 때 사용하는
# welch t-검정을 사용합니다.

# 평균의 차이의 분포이기 때문에 두 평균의 차이가 없다는 귀무가설 하에서는
# 히스토그램의 중심이 0이 됩니다.

# 세로선은 극단적으로 오른쪽 꼬리에 위치합니다.
# 이는 우연히 나타날 수 있는 범위에서 벗어남을 의미합니다.

# 이 세로선(6.27) 오른쪽 넓이의 두 배가 실제 p-값(0.0001)에 해당하며,
# 이 값이 유의수준(0.05)보다 훨씬 작습니다.

# t-검정 결과 A방식이 B방식보다 통계적으로 유의미하게 높은 성적 향상을 보였으므로,
# 교육 기업은 A방식을 주력 상품으로 마케팅하는 한편,
# B방식이 왜 상대적으로 효과가 낮았는지(콘텐츠 구성, 난이도 등) 원인을 분석해
# 개선하거나 폐기 여부를 검토하는 것이 합리적입니다.
```

# 문제 4. 연령대별로 선호하는 영화 장르가 다를까? (난이도: 중)

**📘 문제**

- 한 영화관에서 고객의 **연령대(20대, 30대, 40대)**에 따라 선호하는 **영화 장르(액션, 로맨스)**에 차이가 있는지 궁금해졌습니다.

- 이를 알아보기 위해 300명의 고객을 대상으로 설문조사를 진행하여 분할표를 얻었습니다.

- 이번 실습에서는 **카이제곱 독립성 검정(Chi-squared Test of Independence)**으로
  두 범주형 변수 사이에 연관성이 있는지 검증합니다.

- 또한 **"두 변수가 서로 아무 관련이 없다면(독립이라면)"** 어떤 카이제곱 통계량들이 나올 수 있는지 시뮬레이션으로 확인해 봅니다.

```python
# 데이터 생성 (관측 빈도 분할표)
data = {'액션': [70, 50, 30],
        '로맨스': [30, 60, 60]}
observed = pd.DataFrame(data, index=['20대', '30대', '40대'])

print("관측 빈도 (Observed Frequencies):")
print(observed)

# 분할표 시각화
observed.plot(kind='bar', figsize=(8, 5))
plt.title("연령대별 선호 장르 응답 수")
plt.xlabel("연령대")
plt.ylabel("응답 수 (명)")
plt.xticks(rotation=0)
plt.show()
```

**📌 아래를 수행해 보세요:**

- 귀무가설(H₀)과 대립가설(H₁)을 세워 봅시다.

- 카이제곱 독립성 검정을 수행하여 카이제곱 통계량, p-값, 자유도, **기대 빈도**를 구해봅시다.

- **"연령대와 선호 장르가 독립이라면"** 어떤 분할표들이 나타날 수 있는지 5,000번 시뮬레이션해 봅시다.

- 시뮬레이션으로 얻은 카이제곱 통계량 분포와 실제 관찰된 카이제곱 통계량을 비교해 봅시다.

```python

# [문제 4] Q1. 귀무가설(H₀)과 대립가설(H₁)을 주석으로 작성해봅시다.
# H₀ (귀무가설): 두 변수 (나이와 선호영화장르)는 서로 독립이다.
# H₁ (대립가설): 두 변수는 서로 독립이 아니다.
```

```python

# [문제 4] Q2. 카이제곱 독립성 검정을 수행하고 카이제곱 통계량, p-값, 자유도, 기대 빈도를 구해봅시다.
# 기대 빈도는 관측 빈도와 비교하기 쉽도록 데이터프레임 형태로 출력해보세요.

# 여기에 코드를 작성해주세요.

from scipy import stats

chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)

expected_df = pd.DataFrame(expected, index=observed.index, columns=observed.columns)

print(f"카이제곱 통계량: {chi2_stat:.4f}")
print(f"p-값: {p_value:.6f}")
print(f"자유도: {dof}")
print("\n기대 빈도 (Expected Frequencies):")
print(expected_df.round(2))
```

```python
# [문제 4] Q3. H₀가 사실(두 변수가 독립)이라고 가정하고 시뮬레이션을 5,000번 수행해봅시다.
# 각 반복마다 300명의 가상 고객에게 연령대와 장르를 '독립적으로' 할당한 후,
# 가상 분할표를 만들고 카이제곱 통계량을 계산하여 리스트에 저장해보세요.
# 힌트 1: 전체 비율(observed.sum(axis=0) / 전체, observed.sum(axis=1) / 전체)을 확률로 사용
# 힌트 2: np.random.choice 로 300명의 연령대와 장르를 각각 독립적으로 생성
# 힌트 3: pd.crosstab 으로 분할표 생성 후 stats.chi2_contingency 로 통계량 계산

# 여기에 코드를 작성해주세요.

n_total = observed.values.sum()  # 300명

# 힌트 1: 전체 비율을 확률로 사용
age_probs = observed.sum(axis=1) / n_total       # 20대/30대/40대 각각의 비율
genre_probs = observed.sum(axis=0) / n_total     # 액션/로맨스 각각의 비율

age_labels = observed.index.tolist()       # ['20대', '30대', '40대']
genre_labels = observed.columns.tolist()   # ['액션', '로맨스']

np.random.seed(2025)
sim_chi2_stats = []

for i in range(5000):
    # 힌트 2: 연령대와 장르를 각각 독립적으로 무작위 생성
    sim_ages = np.random.choice(age_labels, size=n_total, p=age_probs.values)
    sim_genres = np.random.choice(genre_labels, size=n_total, p=genre_probs.values)
    
    # 힌트 3: 가상 분할표 생성 후 카이제곱 통계량 계산
    sim_table = pd.crosstab(sim_ages, sim_genres)
    sim_chi2, _, _, _ = stats.chi2_contingency(sim_table)
    
    sim_chi2_stats.append(sim_chi2)

sim_chi2_stats = np.array(sim_chi2_stats)

print(f"시뮬레이션 카이제곱 통계량의 평균: {sim_chi2_stats.mean():.4f} (이론값: 자유도={dof})")
print(f"시뮬레이션 카이제곱 통계량의 표준편차: {sim_chi2_stats.std():.4f}")
```

```python
# [문제 4] Q4. 시뮬레이션으로 얻은 카이제곱 통계량의 분포를 히스토그램으로 그려봅시다.
# 실제 관찰된 카이제곱 통계량을 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.

plt.figure(figsize=(9, 5))
sns.histplot(sim_chi2_stats, bins=40, kde=True)

plt.axvline(chi2_stat, color='red', linestyle='--', linewidth=2,
            label=f'실제 관찰된 카이제곱 통계량: {chi2_stat:.2f}')

plt.title(f"H₀(연령대와 장르는 독립) 가정 하의 카이제곱 통계량 분포 (5,000회 시뮬레이션)")
plt.xlabel("카이제곱 통계량")
plt.legend()
plt.show()
```

# 문제 5. 미니 프로젝트 - 어떤 신규 비료가 가장 효과적일까? (난이도: 상)

**📘 문제**

- 한 농업 연구소에서 새로 개발한 비료 3종류(A, B, C)의 생산량 증대 효과를 비교하고자 합니다.

- 동일한 조건의 밭 30개를 준비하여, 각각 10개씩 비료 A, B, C를 투여한 후 수확량을 측정했습니다.

- 이번 실습에서는 **분산분석(ANOVA)**으로 세 비료 간 평균 수확량에 차이가 있는지 검증하고,
  차이가 있다면 **사후분석(Tukey's HSD)**으로 어떤 비료끼리 차이가 나는지 확인합니다.

- 마지막으로 **"세 비료의 효과가 완전히 똑같다면"** 어떤 F-통계량이 나올 수 있는지 시뮬레이션하고,
  분석 결과를 바탕으로 연구소에 어떤 비료를 추천할지까지 생각해 봅니다.

```python
# 데이터 생성 (비료 종류별 수확량)
np.random.seed(123)
fertilizer_a = np.random.normal(loc=105, scale=10, size=10)
fertilizer_b = np.random.normal(loc=120, scale=12, size=10)
fertilizer_c = np.random.normal(loc=108, scale=9, size=10)

# 데이터프레임으로 변환 (사후분석을 위해)
df = pd.DataFrame({'수확량': np.concatenate([fertilizer_a, fertilizer_b, fertilizer_c]),
                   '비료종류': ['A']*10 + ['B']*10 + ['C']*10})

# 데이터 탐색 시각화
sns.boxplot(x='비료종류', y='수확량', data=df)
plt.title("비료 종류에 따른 수확량 분포")
plt.show()
```

**📌 아래를 수행해 보세요:**

- 세 그룹에 대해 분산분석(ANOVA)을 수행하여 F-통계량과 p-값을 구해봅시다.

- ANOVA 결과가 유의미할 경우에만 **Tukey's HSD 사후분석**을 수행하고 결과를 해석해 봅시다.

- **"세 비료의 효과가 완전히 똑같다면"** 어떤 F-통계량들이 나오는지 5,000번 시뮬레이션해 봅시다.

- 히스토그램을 그리고, 실제 관찰된 F-통계량이 얼마나 극단적인 값인지 확인해 봅시다.

- 분석 결과를 종합하여 연구소에 추천할 비료를 정해 봅시다.

```python

# [문제 5] Q1. 분산분석(ANOVA)을 수행하여 F-통계량과 p-값을 구해봅시다.
# H₀: 세 비료의 평균 수확량은 모두 같다.
# H₁: 적어도 하나 이상의 비료 평균 수확량은 다르다.

# 여기에 코드를 작성해주세요.

from scipy import stats

f_stat, p_value = stats.f_oneway(fertilizer_a, fertilizer_b, fertilizer_c)

print(f"F-통계량: {f_stat:.4f}")
print(f"p-값: {p_value:.6f}")
```

```python
# [문제 5] Q2. ANOVA의 p-값이 유의수준 0.05보다 작을 경우에만
# Tukey's HSD 사후분석을 수행하고 결과를 출력해봅시다.
# 힌트: pairwise_tukeyhsd(endog=수확량, groups=비료종류, alpha=0.05)

# 여기에 코드를 작성해주세요.
# [문제 5] Q2. ANOVA의 p-값이 유의수준 0.05보다 작을 경우에만
# Tukey's HSD 사후분석을 수행하고 결과를 출력해봅시다.

alpha = 0.05

if p_value < alpha:
    print(f"p-값({p_value:.6f}) < 유의수준({alpha}) → 귀무가설(H₀) 기각")
    print("→ 적어도 한 비료의 평균 수확량이 다릅니다. 사후분석을 진행합니다.\n")
    
    from statsmodels.stats.multicomp import pairwise_tukeyhsd
    
    tukey_result = pairwise_tukeyhsd(endog=df['수확량'], groups=df['비료종류'], alpha=0.05)
    print(tukey_result)
else:
    print(f"p-값({p_value:.6f}) >= 유의수준({alpha}) → 귀무가설(H₀) 기각 실패")
    print("→ 세 비료 간 평균 수확량 차이가 통계적으로 유의미하지 않아 사후분석이 필요하지 않습니다.")
```

```python
# [문제 5] Q3. H₀가 사실(세 그룹의 평균이 모두 같다)이라고 가정하고 시뮬레이션을 5,000번 수행해봅시다.
# 모든 데이터를 합친 평균과 표준편차를 가진 H₀ 모집단에서 10개씩 3개의 가상 그룹을 생성하고,
# 이 가상 그룹들로 ANOVA를 수행하여 F-통계량을 리스트에 저장해보세요.

# 여기에 코드를 작성해주세요.

pooled_mean = df['수확량'].mean()
pooled_std = df['수확량'].std(ddof=1)

np.random.seed(2025)
sim_f_stats = []

for i in range(5000):
    sim_a = np.random.normal(loc=pooled_mean, scale=pooled_std, size=10)
    sim_b = np.random.normal(loc=pooled_mean, scale=pooled_std, size=10)
    sim_c = np.random.normal(loc=pooled_mean, scale=pooled_std, size=10)
    
    sim_f, _ = stats.f_oneway(sim_a, sim_b, sim_c)
    sim_f_stats.append(sim_f)

sim_f_stats = np.array(sim_f_stats)

print(f"시뮬레이션 F-통계량의 평균: {sim_f_stats.mean():.4f} (이론값: 약 1)")
print(f"시뮬레이션 F-통계량의 표준편차: {sim_f_stats.std():.4f}")
```

```python
# [문제 5] Q4. 시뮬레이션으로 얻은 F-통계량의 분포를 히스토그램으로 그려봅시다.
# 실제 관찰된 F-통계량을 세로선으로 함께 표시해 봅시다.

# 여기에 코드를 작성해주세요.

plt.figure(figsize=(9, 5))
sns.histplot(sim_f_stats, bins=40, kde=True)

plt.axvline(f_stat, color='red', linestyle='--', linewidth=2,
            label=f'실제 관찰된 F-통계량: {f_stat:.2f}')

plt.title("H₀(세 비료의 평균이 동일) 가정 하의 F-통계량 분포 (5,000회 시뮬레이션)")
plt.xlabel("F-통계량")
plt.legend()
plt.show()
```
